```python
import re

def normalize_text(text):
    if pd.isna(text):
        return ""
    
    text = str(text).lower()
    text = re.sub(r'[^a-z0-9]+', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

# Sirf sample pairs par test
test_norm = sample_pairs.copy()

test_norm["source1_name_norm"] = test_norm["source1_name"].apply(normalize_text)
test_norm["matched_name_norm"] = test_norm["matched_name"].apply(normalize_text)

test_norm["source1_address_norm"] = test_norm["source1_address"].apply(normalize_text)
test_norm["matched_address_norm"] = test_norm["matched_address"].apply(normalize_text)

print(
    test_norm[
        [
            "source1_name",
            "source1_name_norm",
            "matched_name",
            "matched_name_norm",
            "source1_address_norm",
            "matched_address_norm"
        ]
    ].to_string(index=False)
)
```

```python
# Har Source 1 entity ke actual matched IDs ka set banana

true_matches = {}

for _, row in train_ground_truth.dropna(
    subset=["matched_entity_ids"]
).iterrows():

    s1_id = row["source1_entity_id"]

    matched_ids = set(
        row["matched_entity_ids"].split(",")
    )

    true_matches[s1_id] = matched_ids

print("Total Source 1 entities with matches:", len(true_matches))

# Example
example_id = positive_sample.iloc[0]["source1_entity_id"]

print("\nExample Source 1 ID:", example_id)
print("Actual matched IDs:", true_matches.get(example_id, set()))
```

```python
import numpy as np

# S2 + S3 ko ek table mein combine karo
source23 = pd.concat(
    [
        train_source2[["entity_id", "business_name", "business_address", "country"]],
        train_source3[["entity_id", "business_name", "business_address", "country"]]
    ],
    ignore_index=True
)

# Fast ID -> row lookup
source23_ids = source23["entity_id"].values

# Country ke according IDs
country_to_ids = (
    source23.groupby("country")["entity_id"]
    .apply(list)
    .to_dict()
)

rng = np.random.default_rng(42)

negative_pairs = []

for s1_id in positive_sample["source1_entity_id"]:
    
    # Source 1 ki country
    s1_row = train_source1.loc[
        train_source1["entity_id"] == s1_id
    ].iloc[0]
    
    country = s1_row["country"]
    
    # Same country ke candidates
    candidates = country_to_ids.get(country, [])
    
    if not candidates:
        continue
    
    # Random ID choose karo
    while True:
        candidate_id = candidates[
            rng.integers(0, len(candidates))
        ]
        
        # Check karo ki ye actual match na ho
        if candidate_id not in true_matches.get(s1_id, set()):
            break
    
    negative_pairs.append(
        (s1_id, candidate_id)
    )

negative_sample = pd.DataFrame(
    negative_pairs,
    columns=["source1_entity_id", "matched_entity_id"]
)

print("Negative samples:", len(negative_sample))
print("\nFirst 10 negative pairs:")
print(negative_sample.head(10))
```

```python
# Fast negative sampling

# Source 1 ID -> country lookup
s1_country = train_source1.set_index("entity_id")["country"].to_dict()

# Same-country S2/S3 IDs already available in country_to_ids

rng = np.random.default_rng(42)

negative_pairs = []

for s1_id in positive_sample["source1_entity_id"]:

    country = s1_country[s1_id]

    candidates = country_to_ids.get(country, [])

    # Random candidate until it is not a true match
    true_ids = true_matches.get(s1_id, set())

    while True:
        candidate_id = candidates[
            rng.integers(0, len(candidates))
        ]

        if candidate_id not in true_ids:
            break

    negative_pairs.append(
        (s1_id, candidate_id)
    )

negative_sample = pd.DataFrame(
    negative_pairs,
    columns=["source1_entity_id", "matched_entity_id"]
)

print("Negative samples:", len(negative_sample))
print("\nFirst 10 negative pairs:")
print(negative_sample.head(10))
```

```python
# Positive + Negative pairs combine

positive_sample["label"] = 1
negative_sample["label"] = 0

training_pairs = pd.concat(
    [positive_sample, negative_sample],
    ignore_index=True
)

# Shuffle
training_pairs = training_pairs.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("Total training pairs:", len(training_pairs))
print("\nLabel distribution:")
print(training_pairs["label"].value_counts())

print("\nSample:")
print(training_pairs.head(10))
```

```python
from difflib import SequenceMatcher
import re

# S1 aur S2/S3 ka data lookup dictionary
s1_data = train_source1.set_index("entity_id").to_dict("index")
source23_data = source23.set_index("entity_id").to_dict("index")


def normalize_text(text):
    if pd.isna(text):
        return ""
    
    text = str(text).lower()
    text = re.sub(r'\s+', ' ', text).strip()
    return text


def similarity(a, b):
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, a, b).ratio()


def token_similarity(a, b):
    if not a or not b:
        return 0.0
    
    tokens_a = set(a.split())
    tokens_b = set(b.split())
    
    if not tokens_a or not tokens_b:
        return 0.0
    
    intersection = len(tokens_a & tokens_b)
    union = len(tokens_a | tokens_b)
    
    return intersection / union


def number_overlap(a, b):
    nums_a = set(re.findall(r'\d+', a))
    nums_b = set(re.findall(r'\d+', b))
    
    if not nums_a or not nums_b:
        return 0.0
    
    return len(nums_a & nums_b) / len(nums_a | nums_b)


features = []

for _, row in training_pairs.iterrows():

    s1 = s1_data[row["source1_entity_id"]]
    s2 = source23_data[row["matched_entity_id"]]

    name1 = normalize_text(s1["business_name"])
    name2 = normalize_text(s2["business_name"])

    address1 = normalize_text(s1["business_address"])
    address2 = normalize_text(s2["business_address"])

    features.append({
        "name_similarity": similarity(name1, name2),
        "address_similarity": similarity(address1, address2),
        "name_token_similarity": token_similarity(name1, name2),
        "address_token_similarity": token_similarity(address1, address2),
        "number_overlap": number_overlap(address1, address2),
        "country_match": int(s1["country"] == s2["country"]),
        "source1_address_missing": int(not address1),
        "candidate_address_missing": int(not address2),
        "label": row["label"]
    })


training_features = pd.DataFrame(features)

print("Feature shape:", training_features.shape)

print("\nColumns:")
print(training_features.columns.tolist())

print("\nFirst 10 rows:")
print(training_features.head(10))

print("\nLabel distribution:")
print(training_features["label"].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Features aur target
X = training_features.drop(columns=["label"])
y = training_features["label"]

# 80% training, 20% validation
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Validation samples:", len(X_val))

# Random Forest
model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

model.fit(X_train, y_train)

# Validation prediction
y_pred = model.predict(X_val)

print("\nAccuracy:", accuracy_score(y_val, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred))

print("\nClassification Report:")
print(classification_report(y_val, y_pred))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# Positive class probability
val_prob = model.predict_proba(X_val)[:, 1]

for threshold in [0.50, 0.60, 0.70, 0.80, 0.90]:

    pred = (val_prob >= threshold).astype(int)

    precision = precision_score(y_val, pred, zero_division=0)
    recall = recall_score(y_val, pred, zero_division=0)
    f1 = f1_score(y_val, pred, zero_division=0)

    print(
        f"Threshold {threshold:.2f} "
        f"| Precision: {precision:.4f} "
        f"| Recall: {recall:.4f} "
        f"| F1: {f1:.4f}"
    )
```

```python
import re
import pandas as pd

def block_name(text):
    if pd.isna(text):
        return ""
    
    text = str(text).lower()
    text = re.sub(r'[^a-z0-9]+', '', text)
    
    return text[:5]


# Source 1 ke liye block key
train_source1["block_key"] = (
    train_source1["country"].fillna("").astype(str)
    + "_"
    + train_source1["business_name"].apply(block_name)
)

# Source 2 + Source 3
source23 = pd.concat(
    [
        train_source2[["entity_id", "business_name", "business_address", "country"]],
        train_source3[["entity_id", "business_name", "business_address", "country"]]
    ],
    ignore_index=True
)

source23["block_key"] = (
    source23["country"].fillna("").astype(str)
    + "_"
    + source23["business_name"].apply(block_name)
)

# Positive pairs ke liye block key check
check_pairs = positive_sample.copy()

s1_keys = train_source1.set_index("entity_id")["block_key"].to_dict()
s23_keys = source23.set_index("entity_id")["block_key"].to_dict()

check_pairs["s1_block"] = check_pairs["source1_entity_id"].map(s1_keys)
check_pairs["s23_block"] = check_pairs["matched_entity_id"].map(s23_keys)

check_pairs["same_block"] = (
    check_pairs["s1_block"] == check_pairs["s23_block"]
)

print("Total positive samples:", len(check_pairs))
print("Captured by name blocking:", check_pairs["same_block"].sum())
print(
    "Blocking coverage:",
    round(check_pairs["same_block"].mean() * 100, 2),
    "%"
)
```

```python
def block_address(text):
    if pd.isna(text):
        return ""
    
    text = str(text).lower()
    text = re.sub(r'[^a-z0-9]+', '', text)
    
    return text[:6]


# Address block keys
train_source1["address_block"] = (
    train_source1["country"].fillna("").astype(str)
    + "_"
    + train_source1["business_address"].apply(block_address)
)

source23["address_block"] = (
    source23["country"].fillna("").astype(str)
    + "_"
    + source23["business_address"].apply(block_address)
)

# Dictionaries
s1_address_keys = train_source1.set_index("entity_id")["address_block"].to_dict()
s23_address_keys = source23.set_index("entity_id")["address_block"].to_dict()

# Add address blocks
check_pairs["s1_address_block"] = (
    check_pairs["source1_entity_id"].map(s1_address_keys)
)

check_pairs["s23_address_block"] = (
    check_pairs["matched_entity_id"].map(s23_address_keys)
)

# Address match
check_pairs["same_address_block"] = (
    check_pairs["s1_address_block"] == check_pairs["s23_address_block"]
)

# Name OR Address
check_pairs["captured"] = (
    check_pairs["same_block"] |
    check_pairs["same_address_block"]
)

print("Total positive samples:", len(check_pairs))
print("Captured by name + address blocking:", check_pairs["captured"].sum())
print(
    "Blocking coverage:",
    round(check_pairs["captured"].mean() * 100, 2),
    "%"
)
```

```python
def number_block(text):
    if pd.isna(text):
        return ""
    
    nums = re.findall(r'\d+', str(text))
    
    if not nums:
        return ""
    
    return "_".join(nums[:2])


train_source1["number_block"] = (
    train_source1["country"].fillna("").astype(str)
    + "_"
    + train_source1["business_address"].apply(number_block)
)

source23["number_block"] = (
    source23["country"].fillna("").astype(str)
    + "_"
    + source23["business_address"].apply(number_block)
)


s1_number_keys = train_source1.set_index("entity_id")["number_block"].to_dict()
s23_number_keys = source23.set_index("entity_id")["number_block"].to_dict()


check_pairs["s1_number_block"] = (
    check_pairs["source1_entity_id"].map(s1_number_keys)
)

check_pairs["s23_number_block"] = (
    check_pairs["matched_entity_id"].map(s23_number_keys)
)


check_pairs["same_number_block"] = (
    (check_pairs["s1_number_block"] == check_pairs["s23_number_block"])
    & (check_pairs["s1_number_block"] != "")
)


check_pairs["captured_final"] = (
    check_pairs["same_block"]
    | check_pairs["same_address_block"]
    | check_pairs["same_number_block"]
)


print("Total positive samples:", len(check_pairs))
print("Captured by name + address + number:",
      check_pairs["captured_final"].sum())

print(
    "Blocking coverage:",
    round(check_pairs["captured_final"].mean() * 100, 2),
    "%"
)
```

```python
# ==============================
# FAST BLOCKING SETUP
# ==============================

import pandas as pd
import re

def clean_block(text):
    if pd.isna(text):
        return ""
    
    text = str(text).lower()
    text = re.sub(r'[^a-z0-9]+', '', text)
    return text


def name_key(text):
    text = clean_block(text)
    return text[:4]


def address_key(text):
    text = clean_block(text)
    return text[:6]


# TEST Source 1
test_source1["name_key"] = (
    test_source1["country"].fillna("").astype(str)
    + "_"
    + test_source1["business_name"].apply(name_key)
)

test_source1["address_key"] = (
    test_source1["country"].fillna("").astype(str)
    + "_"
    + test_source1["business_address"].apply(address_key)
)


# TEST Source 2 + Source 3
test_source23 = pd.concat(
    [
        test_source2[["entity_id", "business_name", "business_address", "country"]],
        test_source3[["entity_id", "business_name", "business_address", "country"]]
    ],
    ignore_index=True
)

test_source23["name_key"] = (
    test_source23["country"].fillna("").astype(str)
    + "_"
    + test_source23["business_name"].apply(name_key)
)

test_source23["address_key"] = (
    test_source23["country"].fillna("").astype(str)
    + "_"
    + test_source23["business_address"].apply(address_key)
)


print("Test Source 1:", len(test_source1))
print("Test Source 2:", len(test_source2))
print("Test Source 3:", len(test_source3))
print("Total candidates source:", len(test_source23))
```

```python
# ==========================================
# FAST CANDIDATE INDEX
# ==========================================

from collections import defaultdict

name_index = defaultdict(list)
address_index = defaultdict(list)

# Source 2 + Source 3 ko index karo
for row in test_source23[
    ["entity_id", "name_key", "address_key"]
].itertuples(index=False):

    entity_id, nk, ak = row

    if nk:
        name_index[nk].append(entity_id)

    if ak:
        address_index[ak].append(entity_id)

print("Name index keys:", len(name_index))
print("Address index keys:", len(address_index))

print("Index ready.")
```

```python
# ==========================================
# GENERATE TEST CANDIDATES
# ==========================================

candidate_rows = []

for row in test_source1[
    ["entity_id", "name_key", "address_key"]
].itertuples(index=False):

    s1_id, nk, ak = row

    candidates = set()

    # Name based candidates
    if nk in name_index:
        candidates.update(name_index[nk])

    # Address based candidates
    if ak in address_index:
        candidates.update(address_index[ak])

    # Save candidates
    candidate_rows.append(
        (
            s1_id,
            ",".join(candidates)
        )
    )


candidate_pairs = pd.DataFrame(
    candidate_rows,
    columns=["source1_entity_id", "candidate_entity_ids"]
)

print("Source1 entities:", len(candidate_pairs))

print(
    "Entities having candidates:",
    (candidate_pairs["candidate_entity_ids"] != "").sum()
)

print(
    "Entities without candidates:",
    (candidate_pairs["candidate_entity_ids"] == "").sum()
)

print("\nSample:")
print(candidate_pairs.head())
```

```python
# ==========================================
# CHECK BLOCK SIZES - DUCKDB
# ==========================================

def clean_sql(col):
    return f"""
        regexp_replace(
            lower(coalesce(CAST({col} AS VARCHAR), '')),
            '[^a-zA-Z0-9]',
            '',
            'g'
        )
    """

s1_name = clean_sql("business_name")
s23_name = clean_sql("business_name")

query = f"""
WITH
s1 AS (
    SELECT
        entity_id,
        country,
        left({s1_name}, 4) AS name_key
    FROM read_csv_auto('{s1_path}', delim='\\t')
),

s23 AS (
    SELECT
        entity_id,
        country,
        left({s23_name}, 4) AS name_key
    FROM read_csv_auto('{s2_path}', delim='\\t')

    UNION ALL

    SELECT
        entity_id,
        country,
        left({s23_name}, 4) AS name_key
    FROM read_csv_auto('{s3_path}', delim='\\t')
),

blocks AS (
    SELECT
        country || '_' || name_key AS block_key,
        COUNT(*) AS candidate_count
    FROM s23
    WHERE name_key <> ''
    GROUP BY 1
)

SELECT
    COUNT(*) AS total_blocks,
    MAX(candidate_count) AS largest_block,
    AVG(candidate_count) AS average_block,
    COUNT(*) FILTER (WHERE candidate_count > 1000) AS blocks_over_1000,
    COUNT(*) FILTER (WHERE candidate_count > 10000) AS blocks_over_10000
FROM blocks
"""

result = con.execute(query).fetchdf()

print(result)
```

```python
# ==========================================
# SAFE NAME CANDIDATE GENERATION
# ==========================================

# Recreate connection if needed
import duckdb

con = duckdb.connect()

# Give DuckDB a controlled memory limit
con.execute("SET memory_limit='12GB'")
con.execute("SET threads=4")

query = f"""
COPY (
    WITH
    s1 AS (
        SELECT
            entity_id AS source1_entity_id,
            country,
            business_name,
            regexp_replace(
                lower(coalesce(CAST(business_name AS VARCHAR), '')),
                '[^a-zA-Z0-9]',
                '',
                'g'
            ) AS clean_name
        FROM read_csv_auto('{s1_path}', delim='\\t')
    ),

    s23 AS (
        SELECT
            entity_id AS candidate_entity_id,
            country,
            business_name,
            regexp_replace(
                lower(coalesce(CAST(business_name AS VARCHAR), '')),
                '[^a-zA-Z0-9]',
                '',
                'g'
            ) AS clean_name
        FROM read_csv_auto('{s2_path}', delim='\\t')

        UNION ALL

        SELECT
            entity_id AS candidate_entity_id,
            country,
            business_name,
            regexp_replace(
                lower(coalesce(CAST(business_name AS VARCHAR), '')),
                '[^a-zA-Z0-9]',
                '',
                'g'
            ) AS clean_name
        FROM read_csv_auto('{s3_path}', delim='\\t')
    ),

    joined AS (
        SELECT
            s1.source1_entity_id,
            s23.candidate_entity_id,

            levenshtein(
                left(s1.clean_name, 30),
                left(s23.clean_name, 30)
            ) AS name_distance

        FROM s1
        JOIN s23
          ON s1.country = s23.country
         AND left(s1.clean_name, 5) = left(s23.clean_name, 5)

        WHERE s1.clean_name <> ''
          AND s23.clean_name <> ''
    ),

    ranked AS (
        SELECT
            source1_entity_id,
            candidate_entity_id,
            name_distance,

            ROW_NUMBER() OVER (
                PARTITION BY source1_entity_id
                ORDER BY name_distance, candidate_entity_id
            ) AS rn

        FROM joined
    )

    SELECT
        source1_entity_id,
        candidate_entity_id
    FROM ranked
    WHERE rn <= 50
)
TO '/kaggle/working/name_candidates.tsv'
(HEADER, DELIMITER '\\t');
"""

print("Candidate generation started...")
con.execute(query)

print("DONE")
print("File created:")
print("/kaggle/working/name_candidates.tsv")
```

```python
import pandas as pd

# Ground truth ko existing variable se use kar rahe hain
gt = ground_truth.copy()

# Empty / NaN matches hatao
gt_nonempty = gt[
    gt["matched_entity_ids"].notna() &
    (gt["matched_entity_ids"].astype(str).str.strip() != "")
].copy()

print("Total Source1 records:", len(gt))
print("Source1 records having matches:", len(gt_nonempty))

# Multiple matched IDs ko alag-alag pairs mein convert karo
positive_pairs = []

for _, row in gt_nonempty.iterrows():
    source1_id = row["source1_entity_id"]
    matched_ids = str(row["matched_entity_ids"]).split(",")

    for candidate_id in matched_ids:
        candidate_id = candidate_id.strip()
        if candidate_id:
            positive_pairs.append((source1_id, candidate_id))

positive_pairs = pd.DataFrame(
    positive_pairs,
    columns=["source1_entity_id", "candidate_entity_id"]
)

print("Total positive matching pairs:", len(positive_pairs))
print("\nSample:")
print(positive_pairs.head())
```

```python
import pandas as pd
import numpy as np

# Reproducible sampling
np.random.seed(42)

# 5,000 positive pairs
positive_sample = positive_pairs.sample(
    n=5000,
    random_state=42
).copy()

# Source1 IDs
source1_ids = train_s1["entity_id"].astype(str).tolist()

# All S2 + S3 candidate IDs
candidate_ids = pd.concat([
    train_s2["entity_id"].astype(str),
    train_s3["entity_id"].astype(str)
], ignore_index=True)

# Create negative pairs
positive_set = set(
    zip(
        positive_pairs["source1_entity_id"].astype(str),
        positive_pairs["candidate_entity_id"].astype(str)
    )
)

negative_pairs = []

while len(negative_pairs) < 5000:
    s1_id = np.random.choice(source1_ids)
    candidate_id = np.random.choice(candidate_ids)

    if (s1_id, candidate_id) not in positive_set:
        negative_pairs.append((s1_id, candidate_id))

negative_sample = pd.DataFrame(
    negative_pairs,
    columns=["source1_entity_id", "candidate_entity_id"]
)

# Add labels
positive_sample["label"] = 1
negative_sample["label"] = 0

training_pairs = pd.concat(
    [positive_sample, negative_sample],
    ignore_index=True
).sample(frac=1, random_state=42).reset_index(drop=True)

print("Positive samples:", len(positive_sample))
print("Negative samples:", len(negative_sample))
print("Total training pairs:", len(training_pairs))

print("\nLabel distribution:")
print(training_pairs["label"].value_counts())

print("\nSample:")
print(training_pairs.head())
```

```python
# FAST SCORING: BATCH 12 TO 347

import os
import time
import joblib
import pandas as pd
import numpy as np
import re
from rapidfuzz import fuzz

scored_dir = "/kaggle/working/scored_candidates"

# Load model
model_package = joblib.load(
    "/kaggle/working/student_matching_model_package.pkl"
)

final_model = model_package["model"]
model_features = model_package["features"]

def token_jaccard(a, b):
    a_tokens = set(re.findall(r"\w+", a.lower()))
    b_tokens = set(re.findall(r"\w+", b.lower()))

    if not a_tokens and not b_tokens:
        return 1.0
    if not a_tokens or not b_tokens:
        return 0.0

    return len(a_tokens & b_tokens) / len(a_tokens | b_tokens)


def number_overlap(a, b):
    nums_a = set(re.findall(r"\d+", a))
    nums_b = set(re.findall(r"\d+", b))

    if not nums_a or not nums_b:
        return 0.0

    return len(nums_a & nums_b) / len(nums_a | nums_b)


for batch_num in range(12, 348):

    output_file = os.path.join(
        scored_dir,
        f"scored_candidates_batch_{batch_num:03d}.tsv"
    )

    # Skip already completed batches
    if os.path.exists(output_file):
        print(f"Batch {batch_num}: already done, skipping")
        continue

    start = time.time()

    candidate_file = (
        f"/kaggle/working/candidates/"
        f"name_candidates_batch_{batch_num:03d}.tsv"
    )

    candidates = pd.read_csv(
        candidate_file,
        sep="\t",
        dtype=str
    )

    s1_ids = candidates["source1_entity_id"].unique()
    candidate_ids = candidates["candidate_entity_id"].unique()

    # Load only required records
    s1_batch = pd.read_csv(
        test_s1_path,
        sep="\t",
        dtype=str
    )

    s1_batch = s1_batch[
        s1_batch["entity_id"].isin(s1_ids)
    ].copy()

    s2_batch = pd.read_csv(
        test_s2_path,
        sep="\t",
        dtype=str
    )

    s3_batch = pd.read_csv(
        test_s3_path,
        sep="\t",
        dtype=str
    )

    s23_batch = pd.concat(
        [s2_batch, s3_batch],
        ignore_index=True
    )

    s23_batch = s23_batch[
        s23_batch["entity_id"].isin(candidate_ids)
    ].copy()

    # Merge Source 1
    pairs = candidates.merge(
        s1_batch[
            ["entity_id", "business_name",
             "business_address", "country"]
        ],
        left_on="source1_entity_id",
        right_on="entity_id",
        how="left"
    ).drop(columns=["entity_id"])

    # Merge Source 2 + Source 3
    pairs = pairs.merge(
        s23_batch[
            ["entity_id", "business_name",
             "business_address", "country"]
        ].rename(columns={
            "entity_id": "candidate_entity_id",
            "business_name": "candidate_business_name",
            "business_address": "candidate_business_address",
            "country": "candidate_country"
        }),
        on="candidate_entity_id",
        how="left"
    )

    pairs = pairs.rename(columns={
        "business_name": "source1_business_name",
        "business_address": "source1_business_address",
        "country": "source1_country"
    })

    # Fill missing values
    text_cols = [
        "source1_business_name",
        "candidate_business_name",
        "source1_business_address",
        "candidate_business_address",
        "source1_country",
        "candidate_country"
    ]

    for col in text_cols:
        pairs[col] = pairs[col].fillna("").astype(str)

    # Fast similarity features
    pairs["name_similarity"] = [
        fuzz.ratio(a, b) / 100
        for a, b in zip(
            pairs["source1_business_name"],
            pairs["candidate_business_name"]
        )
    ]

    pairs["address_similarity"] = [
        fuzz.ratio(a, b) / 100
        for a, b in zip(
            pairs["source1_business_address"],
            pairs["candidate_business_address"]
        )
    ]

    pairs["name_token_similarity"] = [
        token_jaccard(a, b)
        for a, b in zip(
            pairs["source1_business_name"],
            pairs["candidate_business_name"]
        )
    ]

    pairs["address_token_similarity"] = [
        token_jaccard(a, b)
        for a, b in zip(
            pairs["source1_business_address"],
            pairs["candidate_business_address"]
        )
    ]

    pairs["number_overlap"] = [
        number_overlap(a, b)
        for a, b in zip(
            pairs["source1_business_address"],
            pairs["candidate_business_address"]
        )
    ]

    pairs["country_match"] = (
        pairs["source1_country"].str.lower()
        == pairs["candidate_country"].str.lower()
    ).astype(int)

    pairs["source1_address_missing"] = (
        pairs["source1_business_address"].str.strip() == ""
    ).astype(int)

    pairs["candidate_address_missing"] = (
        pairs["candidate_business_address"].str.strip() == ""
    ).astype(int)

    # Predict
    pairs["match_probability"] = (
        final_model.predict_proba(
            pairs[model_features]
        )[:, 1]
    )

    # Save
    pairs[
        [
            "source1_entity_id",
            "candidate_entity_id",
            "match_probability"
        ]
    ].to_csv(
        output_file,
        sep="\t",
        index=False
    )

    print(
        f"Batch {batch_num:03d} done | "
        f"Rows: {len(pairs):,} | "
        f"Time: {time.time() - start:.1f}s"
    )

print("\nALL REMAINING BATCHES COMPLETED!")
```

```python
import pandas as pd

base = "/kaggle/input/datasets/harshnagayach1908/business-entity-resolution-problem-2-0/dataset/train"

train_s1 = pd.read_csv(f"{base}/train_source1.tsv", sep="\t")
train_s2 = pd.read_csv(f"{base}/train_source2.tsv", sep="\t")
train_s3 = pd.read_csv(f"{base}/train_source3.tsv", sep="\t")
ground_truth = pd.read_csv(f"{base}/train_ground_truth.tsv", sep="\t")

print("Source 1:", train_s1.shape)
print("Source 2:", train_s2.shape)
print("Source 3:", train_s3.shape)
print("Ground Truth:", ground_truth.shape)

print("\nGround Truth columns:")
print(ground_truth.columns.tolist())
```

```python
import pandas as pd
import numpy as np

# Ground truth se valid matches
positive_pairs = []

for _, row in ground_truth.iterrows():
    s1_id = row["source1_entity_id"]
    matched = row["matched_entity_ids"]

    if pd.notna(matched):
        for s23_id in str(matched).split(","):
            positive_pairs.append((s1_id, s23_id.strip()))

positive_pairs = pd.DataFrame(
    positive_pairs,
    columns=["source1_entity_id", "candidate_entity_id"]
)

print("Total positive pairs:", len(positive_pairs))
print("\nSample:")
print(positive_pairs.head())
```

```python
# 5,000 positive pairs
positive_sample = positive_pairs.sample(
    n=5000,
    random_state=42
).copy()

# Source 1 IDs jinke liye true matches already hain
true_match_set = set(
    zip(
        positive_pairs["source1_entity_id"],
        positive_pairs["candidate_entity_id"]
    )
)

# Random negative pairs
rng = np.random.default_rng(42)

s1_ids = train_s1["entity_id"].values
s23_ids = pd.concat([
    train_s2["entity_id"],
    train_s3["entity_id"]
]).values

negative_pairs = []

while len(negative_pairs) < 5000:
    s1_batch = rng.choice(s1_ids, 10000)
    s23_batch = rng.choice(s23_ids, 10000)

    for s1_id, s23_id in zip(s1_batch, s23_batch):
        if (s1_id, s23_id) not in true_match_set:
            negative_pairs.append((s1_id, s23_id))

            if len(negative_pairs) >= 5000:
                break

negative_sample = pd.DataFrame(
    negative_pairs,
    columns=["source1_entity_id", "candidate_entity_id"]
)

# Labels
positive_sample["label"] = 1
negative_sample["label"] = 0

training_pairs = pd.concat(
    [positive_sample, negative_sample],
    ignore_index=True
)

print("Positive pairs:", len(positive_sample))
print("Negative pairs:", len(negative_sample))
print("Total training pairs:", len(training_pairs))

print("\nLabel distribution:")
print(training_pairs["label"].value_counts())
```

```python
from difflib import SequenceMatcher
import re
import pandas as pd
import numpy as np

def normalize_text(text):
    if pd.isna(text):
        return ""
    return re.sub(r"\s+", " ", str(text).lower()).strip()

def token_jaccard(a, b):
    a_tokens = set(normalize_text(a).split())
    b_tokens = set(normalize_text(b).split())

    if not a_tokens and not b_tokens:
        return 1.0
    if not a_tokens or not b_tokens:
        return 0.0

    return len(a_tokens & b_tokens) / len(a_tokens | b_tokens)

def number_overlap(a, b):
    nums_a = set(re.findall(r"\d+", str(a)))
    nums_b = set(re.findall(r"\d+", str(b)))

    if not nums_a or not nums_b:
        return 0.0

    return len(nums_a & nums_b) / len(nums_a | nums_b)

# Required IDs only
s1_ids = set(training_pairs["source1_entity_id"])
s23_ids = set(training_pairs["candidate_entity_id"])

s1_lookup = train_s1[
    train_s1["entity_id"].isin(s1_ids)
].set_index("entity_id")

s23_all = pd.concat([train_s2, train_s3], ignore_index=True)

s23_lookup = s23_all[
    s23_all["entity_id"].isin(s23_ids)
].set_index("entity_id")

print("S1 lookup:", len(s1_lookup))
print("S2/S3 lookup:", len(s23_lookup))

feature_rows = []

for _, row in training_pairs.iterrows():

    s1 = s1_lookup.loc[row["source1_entity_id"]]
    s23 = s23_lookup.loc[row["candidate_entity_id"]]

    name1 = normalize_text(s1["business_name"])
    name2 = normalize_text(s23["business_name"])

    addr1 = normalize_text(s1["business_address"])
    addr2 = normalize_text(s23["business_address"])

    feature_rows.append({
        "name_similarity": SequenceMatcher(None, name1, name2).ratio(),
        "address_similarity": SequenceMatcher(None, addr1, addr2).ratio(),
        "name_token_similarity": token_jaccard(name1, name2),
        "address_token_similarity": token_jaccard(addr1, addr2),
        "number_overlap": number_overlap(addr1, addr2),
        "country_match": int(
            normalize_text(s1["country"]) ==
            normalize_text(s23["country"])
        ),
        "source1_address_missing": int(not addr1),
        "candidate_address_missing": int(not addr2),
        "label": row["label"]
    })

features_df = pd.DataFrame(feature_rows)

print("\nFeature shape:", features_df.shape)
print("\nFeatures:")
print(features_df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
import joblib
import os

# Features and label
features = [
    "name_similarity",
    "address_similarity",
    "name_token_similarity",
    "address_token_similarity",
    "number_overlap",
    "country_match",
    "source1_address_missing",
    "candidate_address_missing"
]

X = features_df[features]
y = features_df["label"]

# Train / validation split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Validation rows:", len(X_test))

# Random Forest
final_model = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("\nTraining model...")
final_model.fit(X_train, y_train)

# Validation
y_pred = final_model.predict(X_test)

print("\n========== CLASSIFICATION REPORT ==========")
print(classification_report(y_test, y_pred, digits=4))

print("========== CONFUSION MATRIX ==========")
print(confusion_matrix(y_test, y_pred))

# Save model + feature list
model_package = {
    "model": final_model,
    "features": features
}

model_path = "/kaggle/working/student_matching_model_package.pkl"

joblib.dump(model_package, model_path)

print("\n✅ MODEL SAVED")
print("File:", model_path)
print(
    "Size:",
    round(os.path.getsize(model_path) / (1024 * 1024), 2),
    "MB"
)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import pandas as pd
import numpy as np
import re
from difflib import SequenceMatcher

# Paths
base = "/kaggle/input/datasets/harshnagayach1908/business-entity-resolution-problem-2-0/dataset/train"

s1_path = base + "/train_source1.tsv"
s2_path = base + "/train_source2.tsv"
s3_path = base + "/train_source3.tsv"
gt_path = base + "/train_ground_truth.tsv"

print("Loading training data...")

s1 = pd.read_csv(s1_path, sep="\t")
s2 = pd.read_csv(s2_path, sep="\t")
s3 = pd.read_csv(s3_path, sep="\t")
gt = pd.read_csv(gt_path, sep="\t")

print("Loaded:")
print("S1:", s1.shape)
print("S2:", s2.shape)
print("S3:", s3.shape)
print("GT:", gt.shape)
```

```python
# Ground truth se positive pairs
true_matches = {}

for _, row in gt.iterrows():
    s1_id = row["source1_entity_id"]
    matched = row["matched_entity_ids"]

    if pd.notna(matched) and str(matched).strip():
        ids = str(matched).split(",")
        true_matches[s1_id] = set(x.strip() for x in ids)

# Positive pairs
positive_pairs = []

for s1_id, candidate_ids in true_matches.items():
    for cid in candidate_ids:
        positive_pairs.append((s1_id, cid))

positive_pairs = pd.DataFrame(
    positive_pairs,
    columns=["source1_entity_id", "candidate_entity_id"]
)

print("Positive pairs:", len(positive_pairs))

# 5,000 positive
positive_sample = positive_pairs.sample(
    n=min(5000, len(positive_pairs)),
    random_state=42
)

# Candidate lookup
s1_dict = s1.set_index("entity_id").to_dict("index")

s23 = pd.concat([s2, s3], ignore_index=True)
s23_dict = s23.set_index("entity_id").to_dict("index")

# Negative pairs
rng = np.random.default_rng(42)

negative_pairs = []

s1_ids = list(s1_dict.keys())
s23_ids = list(s23_dict.keys())

positive_set = set(
    (r["source1_entity_id"], r["candidate_entity_id"])
    for _, r in positive_pairs.iterrows()
)

while len(negative_pairs) < 5000:
    a = rng.choice(s1_ids)
    b = rng.choice(s23_ids)

    if (a, b) not in positive_set:
        negative_pairs.append((a, b))

negative_sample = pd.DataFrame(
    negative_pairs,
    columns=["source1_entity_id", "candidate_entity_id"]
)

print("Positive sample:", len(positive_sample))
print("Negative sample:", len(negative_sample))
```

```python
# FAST SAMPLING — no slow while loop

import pandas as pd
import numpy as np

# 5,000 positive pairs
positive_sample = positive_pairs.sample(
    n=5000,
    random_state=42
).reset_index(drop=True)

# IDs as numpy arrays
s1_ids = s1["entity_id"].dropna().astype(str).values
s23_ids = pd.concat([
    s2["entity_id"],
    s3["entity_id"]
]).dropna().astype(str).values

# Positive pair keys
positive_keys = set(
    positive_sample["source1_entity_id"].astype(str)
    + "||" +
    positive_sample["candidate_entity_id"].astype(str)
)

# Generate a large random pool ONCE
rng = np.random.default_rng(42)

neg_s1 = rng.choice(s1_ids, size=20000)
neg_s23 = rng.choice(s23_ids, size=20000)

negative_sample = pd.DataFrame({
    "source1_entity_id": neg_s1,
    "candidate_entity_id": neg_s23
})

# Remove actual positive pairs
negative_sample["key"] = (
    negative_sample["source1_entity_id"]
    + "||" +
    negative_sample["candidate_entity_id"]
)

negative_sample = negative_sample[
    ~negative_sample["key"].isin(positive_keys)
].drop(columns="key").drop_duplicates()

# Take 5,000
negative_sample = negative_sample.head(5000).reset_index(drop=True)

print("Positive sample:", len(positive_sample))
print("Negative sample:", len(negative_sample))

# Combine
train_pairs = pd.concat(
    [positive_sample, negative_sample],
    ignore_index=True
)

train_pairs["label"] = (
    [1] * len(positive_sample) +
    [0] * len(negative_sample)
)

print("Total training pairs:", len(train_pairs))
print("\nLabel distribution:")
print(train_pairs["label"].value_counts())
```

```python
from difflib import SequenceMatcher
import re
import pandas as pd
import numpy as np

def normalize_text(x):
    if pd.isna(x):
        return ""
    return re.sub(r"\s+", " ", str(x).lower()).strip()

def token_jaccard(a, b):
    a = set(normalize_text(a).split())
    b = set(normalize_text(b).split())

    if not a and not b:
        return 1.0
    if not a or not b:
        return 0.0

    return len(a & b) / len(a | b)

def number_overlap(a, b):
    nums_a = set(re.findall(r"\d+", str(a)))
    nums_b = set(re.findall(r"\d+", str(b)))

    if not nums_a or not nums_b:
        return 0.0

    return len(nums_a & nums_b) / len(nums_a | nums_b)

# Small lookup tables
s1_lookup = s1.set_index("entity_id")
s23_lookup = pd.concat([s2, s3]).set_index("entity_id")

features = []

for _, row in train_pairs.iterrows():

    a = s1_lookup.loc[row["source1_entity_id"]]
    b = s23_lookup.loc[row["candidate_entity_id"]]

    name1 = normalize_text(a["business_name"])
    name2 = normalize_text(b["business_name"])

    addr1 = normalize_text(a["business_address"])
    addr2 = normalize_text(b["business_address"])

    features.append([
        SequenceMatcher(None, name1, name2).ratio(),
        SequenceMatcher(None, addr1, addr2).ratio(),
        token_jaccard(name1, name2),
        token_jaccard(addr1, addr2),
        number_overlap(addr1, addr2),
        int(str(a["country"]).lower() == str(b["country"]).lower()),
        int(not addr1),
        int(not addr2),
        row["label"]
    ])

feature_cols = [
    "name_similarity",
    "address_similarity",
    "name_token_similarity",
    "address_token_similarity",
    "number_overlap",
    "country_match",
    "source1_address_missing",
    "candidate_address_missing",
    "label"
]

features = pd.DataFrame(features, columns=feature_cols)

print("Feature shape:", features.shape)
print(features.head())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import joblib

X = features.drop(columns=["label"])
y = features["label"]

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

final_model = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Training model...")

final_model.fit(X_train, y_train)

pred = final_model.predict(X_val)

print("\nValidation Results:")
print(classification_report(y_val, pred))
print("Confusion Matrix:")
print(confusion_matrix(y_val, pred))

# Save model
model_package = {
    "model": final_model,
    "features": list(X.columns)
}

model_path = "/kaggle/working/student_matching_model_package.pkl"

joblib.dump(model_package, model_path)

print("\n✅ Model saved:", model_path)
```

```python
import os
import time
import pandas as pd
import duckdb

# Final output folder
final_dir = "/kaggle/working/output"
os.makedirs(final_dir, exist_ok=True)

candidate_out = final_dir + "/candidate_pairs.tsv"
matching_out = final_dir + "/matching_results.tsv"

# Remove old partial files if they exist
for f in [candidate_out, matching_out]:
    if os.path.exists(f):
        os.remove(f)

# Count Source1
total_s1 = con.execute(f"""
    SELECT COUNT(*)
    FROM read_csv_auto('{test_s1_path}', delim='\t')
""").fetchone()[0]

batch_size = 20000
total_batches = (total_s1 + batch_size - 1) // batch_size

print("Total Source1:", total_s1)
print("Batch size:", batch_size)
print("Total batches:", total_batches)

# Create headers
pd.DataFrame(columns=[
    "source1_entity_id", "candidate_entity_ids"
]).to_csv(
    candidate_out, sep="\t", index=False
)

pd.DataFrame(columns=[
    "source1_entity_id", "matched_entity_ids"
]).to_csv(
    matching_out, sep="\t", index=False
)

start_time = time.time()

for batch_no in range(total_batches):

    offset = batch_no * batch_size

    query = f"""
    WITH s1 AS (
        SELECT
            entity_id AS source1_entity_id,
            business_name,
            country,
            regexp_replace(
                lower(coalesce(business_name, '')),
                '[^a-zA-Z0-9]', '', 'g'
            ) AS clean_name
        FROM read_csv_auto('{test_s1_path}', delim='\t')
        LIMIT {batch_size} OFFSET {offset}
    ),

    joined AS (
        SELECT
            s1.source1_entity_id,
            s23.candidate_entity_id,
            levenshtein(
                left(s1.clean_name, 30),
                left(s23.clean_name, 30)
            ) AS edit_distance,
            length(s1.clean_name) AS name_len
        FROM s1
        JOIN s23_index s23
          ON s1.country = s23.country
         AND length(s1.clean_name) >= 5
         AND length(s23.clean_name) >= 5
         AND left(s1.clean_name, 5) = left(s23.clean_name, 5)
    ),

    ranked AS (
        SELECT
            *,
            row_number() OVER (
                PARTITION BY source1_entity_id
                ORDER BY edit_distance
            ) AS rn
        FROM joined
    )

    SELECT
        source1_entity_id,
        candidate_entity_id,
        edit_distance,
        name_len
    FROM ranked
    WHERE rn <= 5
    """

    df = con.execute(query).df()

    if len(df) > 0:

        # Candidate file: all top-5 candidates
        candidate_df = (
            df.groupby("source1_entity_id")["candidate_entity_id"]
            .apply(lambda x: ",".join(x.astype(str)))
            .reset_index()
        )

        candidate_df.columns = [
            "source1_entity_id",
            "candidate_entity_ids"
        ]

        candidate_df.to_csv(
            candidate_out,
            sep="\t",
            index=False,
            mode="a",
            header=False
        )

        # Matching:
        # Accept closest candidate when name similarity >= 90%
        df["similarity"] = 1 - (
            df["edit_distance"] /
            df["name_len"].clip(lower=1)
        )

        best = (
            df.sort_values(
                ["source1_entity_id", "edit_distance"]
            )
            .groupby("source1_entity_id")
            .first()
            .reset_index()
        )

        best["matched_entity_ids"] = best.apply(
            lambda r: (
                str(r["candidate_entity_id"])
                if r["similarity"] >= 0.90
                else ""
            ),
            axis=1
        )

        matching_df = best[
            ["source1_entity_id", "matched_entity_ids"]
        ]

        matching_df.to_csv(
            matching_out,
            sep="\t",
            index=False,
            mode="a",
            header=False
        )

    print(
        f"Batch {batch_no+1}/{total_batches} done | "
        f"Rows: {len(df)} | "
        f"Time: {time.time()-start_time:.1f}s"
    )

print("\n✅ FINAL FILES CREATED")
print(candidate_out)
print(matching_out)

print("\nCandidate size:",
      os.path.getsize(candidate_out) / (1024*1024), "MB")

print("Matching size:",
      os.path.getsize(matching_out) / (1024*1024), "MB")
```

```python
import os
import time

candidate_out = "/kaggle/working/output/candidate_pairs.tsv"
matching_out = "/kaggle/working/output/matching_results.tsv"

# Start fresh
for f in [candidate_out, matching_out]:
    if os.path.exists(f):
        os.remove(f)

batch_size = 50000
total_s1 = 1732544
total_batches = (total_s1 + batch_size - 1) // batch_size

print("Total Source1:", total_s1)
print("Batch size:", batch_size)
print("Total batches:", total_batches)

start = time.time()

for b in range(total_batches):

    offset = b * batch_size

    # -----------------------------
    # Candidate pairs
    # -----------------------------
    candidate_query = f"""
    WITH s1 AS (
        SELECT
            entity_id AS source1_entity_id,
            country,
            regexp_replace(
                lower(coalesce(business_name, '')),
                '[^a-zA-Z0-9]', '', 'g'
            ) AS clean_name
        FROM read_csv_auto('{test_s1_path}', delim='\t')
        LIMIT {batch_size} OFFSET {offset}
    ),

    pairs AS (
        SELECT
            s1.source1_entity_id,
            s23.candidate_entity_id
        FROM s1
        JOIN s23_index s23
          ON s1.country = s23.country
         AND length(s1.clean_name) >= 5
         AND length(s23.clean_name) >= 5
         AND left(s1.clean_name, 5) = left(s23.clean_name, 5)
    ),

    grouped AS (
        SELECT
            source1_entity_id,
            list_slice(
                list_sort(list(candidate_entity_id)),
                1,
                5
            ) AS candidates
        FROM pairs
        GROUP BY source1_entity_id
    )

    SELECT
        s1.source1_entity_id,
        coalesce(
            array_to_string(g.candidates, ','),
            ''
        ) AS candidate_entity_ids
    FROM s1
    LEFT JOIN grouped g
      ON s1.source1_entity_id = g.source1_entity_id
    """

    cand_df = con.execute(candidate_query).df()

    cand_df.to_csv(
        candidate_out,
        sep="\t",
        index=False,
        mode="a",
        header=(b == 0)
    )

    # -----------------------------
    # Exact matches
    # -----------------------------
    match_query = f"""
    WITH s1 AS (
        SELECT
            entity_id AS source1_entity_id,
            country,
            regexp_replace(
                lower(coalesce(business_name, '')),
                '[^a-zA-Z0-9]', '', 'g'
            ) AS clean_name
        FROM read_csv_auto('{test_s1_path}', delim='\t')
        LIMIT {batch_size} OFFSET {offset}
    ),

    exact AS (
        SELECT
            s1.source1_entity_id,
            s23.candidate_entity_id
        FROM s1
        JOIN s23_index s23
          ON s1.country = s23.country
         AND length(s1.clean_name) >= 5
         AND s1.clean_name = s23.clean_name
    ),

    grouped AS (
        SELECT
            source1_entity_id,
            list_slice(
                list_sort(list(candidate_entity_id)),
                1,
                5
            ) AS matches
        FROM exact
        GROUP BY source1_entity_id
    )

    SELECT
        s1.source1_entity_id,
        coalesce(
            array_to_string(g.matches, ','),
            ''
        ) AS matched_entity_ids
    FROM s1
    LEFT JOIN grouped g
      ON s1.source1_entity_id = g.source1_entity_id
    """

    match_df = con.execute(match_query).df()

    match_df.to_csv(
        matching_out,
        sep="\t",
        index=False,
        mode="a",
        header=(b == 0)
    )

    elapsed = time.time() - start

    print(
        f"Batch {b+1}/{total_batches} done | "
        f"Candidates: {len(cand_df)} | "
        f"Matches: {len(match_df)} | "
        f"Time: {elapsed/60:.1f} min"
    )

print("\n====================================")
print("✅ FINAL FILES CREATED")
print("====================================")
print(candidate_out)
print(matching_out)

print(
    "Candidate size:",
    round(os.path.getsize(candidate_out)/(1024*1024), 2),
    "MB"
)

print(
    "Matching size:",
    round(os.path.getsize(matching_out)/(1024*1024), 2),
    "MB"
)
```

```python
import pandas as pd
import numpy as np
from difflib import SequenceMatcher
import re

def similarity(a, b):
    a = "" if pd.isna(a) else str(a)
    b = "" if pd.isna(b) else str(b)
    return SequenceMatcher(None, a, b).ratio()

def token_similarity(a, b):
    a = "" if pd.isna(a) else str(a)
    b = "" if pd.isna(b) else str(b)

    ta = set(a.lower().split())
    tb = set(b.lower().split())

    if not ta or not tb:
        return 0.0

    return len(ta & tb) / len(ta | tb)

def number_overlap(a, b):
    a = "" if pd.isna(a) else str(a)
    b = "" if pd.isna(b) else str(b)

    na = set(re.findall(r'\d+', a))
    nb = set(re.findall(r'\d+', b))

    if not na or not nb:
        return 0.0

    return len(na & nb) / len(na | nb)

print("Functions ready")
```

```python
import pandas as pd
import numpy as np
from difflib import SequenceMatcher
import re

def similarity(a, b):
    a = "" if pd.isna(a) else str(a)
    b = "" if pd.isna(b) else str(b)
    return SequenceMatcher(None, a, b).ratio()

def token_similarity(a, b):
    a = "" if pd.isna(a) else str(a)
    b = "" if pd.isna(b) else str(b)

    ta = set(a.lower().split())
    tb = set(b.lower().split())

    if not ta or not tb:
        return 0.0

    return len(ta & tb) / len(ta | tb)

def number_overlap(a, b):
    a = "" if pd.isna(a) else str(a)
    b = "" if pd.isna(b) else str(b)

    na = set(re.findall(r'\d+', a))
    nb = set(re.findall(r'\d+', b))

    if not na or not nb:
        return 0.0

    return len(na & nb) / len(na | nb)
```

```python
# ===== BATCH PROCESSOR =====

BATCH_SIZE = 10000
START_OFFSET = 10000
END_OFFSET = 60000

for offset in range(START_OFFSET, END_OFFSET, BATCH_SIZE):

    print(f"\n========== Batch offset {offset} ==========")

    # 1. Get batch
    batch_s1 = con.execute(f"""
        SELECT *
        FROM unresolved_s1
        LIMIT {BATCH_SIZE}
        OFFSET {offset}
    """).fetchdf()

    if len(batch_s1) == 0:
        print("No more rows.")
        break

    # 2. Register batch
    con.register("current_batch", batch_s1)

    # 3. Generate candidates
    fuzzy_test = con.execute("""
        WITH candidates AS (
            SELECT
                s.source1_entity_id,
                c.candidate_entity_id,
                abs(length(s.clean_name) - length(c.clean_name)) AS len_diff
            FROM current_batch s
            JOIN s23_index c
              ON s.country = c.country
             AND left(s.clean_name, 3) = left(c.clean_name, 3)
             AND right(s.clean_name, 3) = right(c.clean_name, 3)
             AND abs(length(s.clean_name) - length(c.clean_name)) <= 2
            WHERE length(s.clean_name) >= 6
              AND length(c.clean_name) >= 6
        ),
        ranked AS (
            SELECT *,
                   ROW_NUMBER() OVER (
                       PARTITION BY source1_entity_id
                       ORDER BY len_diff, candidate_entity_id
                   ) AS rn
            FROM candidates
        )
        SELECT *
        FROM ranked
        WHERE rn <= 20
    """).fetchdf()

    print("Candidate rows:", len(fuzzy_test))

    if len(fuzzy_test) == 0:
        continue

    # 4. Get candidate details
    con.register("current_candidates", fuzzy_test)

    candidate_df = con.execute("""
        SELECT
            f.source1_entity_id,
            s.business_name AS source1_name,
            s.business_address AS source1_address,
            s.country AS source1_country,

            f.candidate_entity_id,
            c.business_name AS candidate_name,
            c.business_address AS candidate_address,
            c.country AS candidate_country

        FROM current_candidates f
        JOIN current_batch s
          ON f.source1_entity_id = s.source1_entity_id
        JOIN s23_index c
          ON f.candidate_entity_id = c.candidate_entity_id
    """).fetchdf()

    # 5. Features
    candidate_df["name_similarity"] = [
        similarity(a, b)
        for a, b in zip(
            candidate_df["source1_name"],
            candidate_df["candidate_name"]
        )
    ]

    candidate_df["address_similarity"] = [
        similarity(a, b)
        for a, b in zip(
            candidate_df["source1_address"],
            candidate_df["candidate_address"]
        )
    ]

    candidate_df["name_token_similarity"] = [
        token_similarity(a, b)
        for a, b in zip(
            candidate_df["source1_name"],
            candidate_df["candidate_name"]
        )
    ]

    candidate_df["address_token_similarity"] = [
        token_similarity(a, b)
        for a, b in zip(
            candidate_df["source1_address"],
            candidate_df["candidate_address"]
        )
    ]

    candidate_df["number_overlap"] = [
        number_overlap(a, b)
        for a, b in zip(
            candidate_df["source1_address"],
            candidate_df["candidate_address"]
        )
    ]

    candidate_df["country_match"] = (
        candidate_df["source1_country"]
        == candidate_df["candidate_country"]
    ).astype(int)

    candidate_df["source1_address_missing"] = (
        candidate_df["source1_address"]
        .fillna("")
        .str.strip() == ""
    ).astype(int)

    candidate_df["candidate_address_missing"] = (
        candidate_df["candidate_address"]
        .fillna("")
        .str.strip() == ""
    ).astype(int)

    # 6. Model
    candidate_df["match_probability"] = model.predict_proba(
        candidate_df[model_features]
    )[:, 1]

    # 7. Strict rule
    strict_rule = candidate_df[
        (candidate_df["match_probability"] >= 0.95) &
        (
            (
                (candidate_df["name_similarity"] >= 0.95) &
                (
                    (candidate_df["address_similarity"] >= 0.35) |
                    (candidate_df["address_token_similarity"] >= 0.50) |
                    (candidate_df["number_overlap"] >= 0.50)
                )
            )
            |
            (
                (candidate_df["name_similarity"] >= 0.90) &
                (
                    (candidate_df["address_similarity"] >= 0.55) |
                    (candidate_df["address_token_similarity"] >= 0.65) |
                    (candidate_df["number_overlap"] >= 0.75)
                )
            )
            |
            (
                (candidate_df["name_similarity"] >= 0.85) &
                (candidate_df["name_similarity"] < 0.90) &
                (
                    (candidate_df["address_similarity"] >= 0.75) |
                    (candidate_df["address_token_similarity"] >= 0.80) |
                    (candidate_df["number_overlap"] >= 0.75)
                )
            )
        )
    ].copy()

    # 8. Best match per Source1
    strict_best = (
        strict_rule
        .sort_values(
            [
                "source1_entity_id",
                "name_similarity",
                "address_similarity",
                "address_token_similarity",
                "number_overlap",
                "match_probability"
            ],
            ascending=[True, False, False, False, False, False]
        )
        .drop_duplicates("source1_entity_id")
    )

    print("Final matches:", len(strict_best))

    # 9. Save
    if len(strict_best) > 0:

        batch_result = strict_best[
            [
                "source1_entity_id",
                "candidate_entity_id",
                "match_probability",
                "name_similarity",
                "address_similarity",
                "address_token_similarity",
                "number_overlap"
            ]
        ].copy()

        con.register("batch_result_df", batch_result)

        con.execute("""
            INSERT INTO fuzzy_matches
            SELECT
                source1_entity_id,
                candidate_entity_id,
                match_probability,
                name_similarity,
                address_similarity,
                address_token_similarity,
                number_overlap
            FROM batch_result_df
        """)

    print(
        "Total saved:",
        con.execute("SELECT COUNT(*) FROM fuzzy_matches").fetchone()[0]
    )

print("\n===== 5 BATCHES COMPLETE =====")
```

```python
output_dir = "/kaggle/working/output"

# ===== FINAL MATCHING RESULTS =====

con.execute(f"""
COPY (
    WITH dedup AS (
        SELECT DISTINCT
            source1_entity_id,
            candidate_entity_id
        FROM improved_matches_v2
    ),
    agg AS (
        SELECT
            source1_entity_id,
            STRING_AGG(
                candidate_entity_id,
                ',' ORDER BY candidate_entity_id
            ) AS matched_entity_ids
        FROM dedup
        GROUP BY source1_entity_id
    )
    SELECT
        s.source1_entity_id,
        COALESCE(a.matched_entity_ids, '') AS matched_entity_ids
    FROM s1_full s
    LEFT JOIN agg a
        ON s.source1_entity_id = a.source1_entity_id
    ORDER BY s.source1_entity_id
)
TO '{output_dir}/matching_results_v2.tsv'
WITH (
    FORMAT CSV,
    DELIMITER '\t',
    HEADER TRUE,
    QUOTE ''
)
""")

# ===== CANDIDATE FILE =====

con.execute(f"""
COPY (
    WITH dedup AS (
        SELECT DISTINCT
            source1_entity_id,
            candidate_entity_id
        FROM improved_matches_v2
    ),
    agg AS (
        SELECT
            source1_entity_id,
            STRING_AGG(
                candidate_entity_id,
                ',' ORDER BY candidate_entity_id
            ) AS candidate_entity_ids
        FROM dedup
        GROUP BY source1_entity_id
    )
    SELECT
        s.source1_entity_id,
        COALESCE(a.candidate_entity_ids, '') AS candidate_entity_ids
    FROM s1_full s
    LEFT JOIN agg a
        ON s.source1_entity_id = a.source1_entity_id
    ORDER BY s.source1_entity_id
)
TO '{output_dir}/candidate_pairs_v2.tsv'
WITH (
    FORMAT CSV,
    DELIMITER '\t',
    HEADER TRUE,
    QUOTE ''
)
""")

import os

for f in [
    f"{output_dir}/matching_results_v2.tsv",
    f"{output_dir}/candidate_pairs_v2.tsv"
]:
    print(f)
    print(round(os.path.getsize(f)/(1024**2), 2), "MB")
```
